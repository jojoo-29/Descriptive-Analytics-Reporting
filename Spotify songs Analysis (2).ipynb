```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")

# 1. LOAD FULL DATASET & EXTRACT RANDOM SAMPLE (N=80)

df_full = pd.read_csv('spotify_songs.csv')
df = df_full.sample(n=80, random_state=42).reset_index(drop=True)
```

```python
# 2. Handle missing values (drop any row with missing data)
df = df.dropna().reset_index(drop=True)

# 3. Selecting ONLY important columns for the project report
important_cols = [
    'track_name', 
    'track_artist', 
    'track_popularity', 
    'playlist_genre', 
    'track_album_release_date',
    'danceability', 
    'energy', 
    'loudness', 
    'valence', 
    'acousticness', 
    'tempo', 
    'duration_ms'
]

df_clean = df[important_cols].copy()
```

```python
# 4. Clean & Convert Date & Duration
# Extract 4-digit year from release date
df_clean['release_year'] = pd.to_datetime(df_clean['track_album_release_date'], errors='coerce').dt.year.astype('Int64')
df_clean.drop(columns=['track_album_release_date'], inplace=True)

# Convert duration from milliseconds to minutes (rounded to 2 decimal places)
df_clean['duration_min'] = (df_clean['duration_ms'] / 60000).round(2)
df_clean.drop(columns=['duration_ms'], inplace=True)

# Strip any extra spaces from text columns
df_clean['track_name'] = df_clean['track_name'].str.strip()
df_clean['track_artist'] = df_clean['track_artist'].str.strip()

# 5. Save the final cleaned table
df_clean.to_csv('spotify_cleaned_sample.csv', index=False)

print("Data Cleaning Complete! Here is the slimmed-down dataset:")
df_clean.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
import os
import shutil

# 1. Load the cleaned CSV
df_clean = pd.read_csv('spotify_cleaned_sample.csv')

# 2. Save as Excel XLSX File
xlsx_filename = "spotify_cleaned_sample.xlsx"
df_clean.to_excel(xlsx_filename, index=False)
print("✅ Saved as XLSX Excel file successfully!")

# 3. Save as Multi-page PDF Table
pdf_filename = "spotify_cleaned_dataset.pdf"
rows_per_page = 25
num_pages = (len(df_clean) + rows_per_page - 1) // rows_per_page

with PdfPages(pdf_filename) as pdf:
    for page in range(num_pages):
        start_idx = page * rows_per_page
        end_idx = min((page + 1) * rows_per_page, len(df_clean))
        df_page = df_clean.iloc[start_idx:end_idx]
        
        fig, ax = plt.subplots(figsize=(14, 8.5))
        ax.axis('tight')
        ax.axis('off')
        
        table_data = [df_clean.columns.tolist()] + df_page.values.tolist()
        table = ax.table(cellText=table_data, cellLoc='center', loc='center')
        table.auto_set_font_size(False)
        table.set_fontsize(6.5)
        table.scale(1, 1.2)
        
        for (r, c), cell in table.get_celld().items():
            if r == 0:
                cell.set_facecolor('#4c72b0')
                cell.set_text_props(color='white', weight='bold', fontsize=7)
            elif r % 2 == 0:
                cell.set_facecolor('#f2f2f2')
                
        plt.title(f"Spotify Cleaned Dataset (Page {page + 1} of {num_pages})", fontsize=11, fontweight='bold', pad=12)
        pdf.savefig(fig, bbox_inches='tight')
        plt.close()

print("✅ Saved as PDF document successfully!")

# 4. Copy both XLSX and PDF directly to your Desktop (For local Jupyter Notebook)
desktop_path = os.path.join(os.path.expanduser('~'), 'Desktop')

if os.path.exists(desktop_path):
    shutil.copy(xlsx_filename, os.path.join(desktop_path, xlsx_filename))
    shutil.copy(pdf_filename, os.path.join(desktop_path, pdf_filename))
    print(f"\n🎉 SUCCESS: Both '{xlsx_filename}' and '{pdf_filename}' have been saved directly to your Desktop!")
else:
    print(f"\n📁 Files generated in your project folder! (Desktop folder not found)")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
import os
import shutil

# 1. Load Excel file
df = pd.read_excel('spotify_cleaned_sample.xlsx')

# 2. Fill any remaining empty years with a default or convert safely to nullable Int64
df['release_year'] = df['release_year'].fillna(2020).astype('Int64')

# 3. Save clean Excel XLSX File
xlsx_filename = "spotify_cleaned_sample.xlsx"
df.to_excel(xlsx_filename, index=False)
print("✅ Saved Excel sheet successfully!")

# 4. Save Multi-page PDF Table (Format years as plain 4-digit text)
pdf_filename = "spotify_cleaned_dataset.pdf"
rows_per_page = 22
num_pages = (len(df) + rows_per_page - 1) // rows_per_page

df_pdf = df.copy()

# Force release_year to display cleanly as plain text without floats (.0)
df_pdf['release_year'] = df_pdf['release_year'].astype(str).str.replace('.0', '', regex=False)

# Truncate long text so cell borders stay neat
df_pdf['track_name'] = df_pdf['track_name'].apply(lambda x: str(x)[:22] + '...' if len(str(x)) > 25 else str(x))
df_pdf['track_artist'] = df_pdf['track_artist'].apply(lambda x: str(x)[:18] + '...' if len(str(x)) > 20 else str(x))

with PdfPages(pdf_filename) as pdf:
    for page in range(num_pages):
        start_idx = page * rows_per_page
        end_idx = min((page + 1) * rows_per_page, len(df_pdf))
        df_page = df_pdf.iloc[start_idx:end_idx]
        
        fig, ax = plt.subplots(figsize=(16, 9))
        ax.axis('tight')
        ax.axis('off')
        
        headers = [col.replace('_', ' ').title() for col in df_pdf.columns]
        table_data = [headers] + df_page.values.tolist()
        
        table = ax.table(cellText=table_data, cellLoc='center', loc='center')
        table.auto_set_font_size(False)
        table.set_fontsize(6)
        table.scale(1, 1.25)
        
        col_widths = [0.14, 0.12, 0.07, 0.06, 0.07, 0.05, 0.06, 0.05, 0.07, 0.06, 0.06, 0.06]
        
        for (r, c), cell in table.get_celld().items():
            cell.set_width(col_widths[c])
            if r == 0:
                cell.set_facecolor('#2c3e50')
                cell.set_text_props(color='white', weight='bold', fontsize=6.5)
            elif r % 2 == 0:
                cell.set_facecolor('#f8f9fa')
            else:
                cell.set_facecolor('#ffffff')
                
        plt.title(f"Spotify Cleaned Dataset — Page {page + 1} of {num_pages}", fontsize=11, fontweight='bold', pad=15)
        pdf.savefig(fig, bbox_inches='tight', pad_inches=0.5)
        plt.close()

print("✅ Saved PDF document successfully without float years!")

# 5. Copy directly to Desktop
desktop_path = os.path.join(os.path.expanduser('~'), 'Desktop')
if os.path.exists(desktop_path):
    shutil.copy(xlsx_filename, os.path.join(desktop_path, xlsx_filename))
    shutil.copy(pdf_filename, os.path.join(desktop_path, pdf_filename))
    print(f"\n🎉 SUCCESS: Both '{xlsx_filename}' and '{pdf_filename}' have been saved directly to your Desktop!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load data & aggregate by year
df = pd.read_excel('spotify_cleaned_sample.xlsx')
year_df = df.groupby('release_year')[['track_popularity', 'duration_min']].mean().reset_index()

# 2. Set dark mode theme
plt.style.use('dark_background')
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 4.8), sharex=True, dpi=300)

# 3. Plot Popularity Area Chart (Neon Cyan)
ax1.plot(year_df['release_year'], year_df['track_popularity'], color='#00F5D4', marker='o', linewidth=2.2, markersize=5)
ax1.fill_between(year_df['release_year'], year_df['track_popularity'], color='#00F5D4', alpha=0.25)
ax1.set_title('Average Song Popularity Over Release Years', fontsize=10, fontweight='bold', color='white', pad=8)
ax1.set_ylabel('Popularity Score', fontsize=8.5, color='#00F5D4', fontweight='bold')
ax1.grid(True, linestyle=':', alpha=0.25, color='gray')
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)

# 4. Plot Song Duration Area Chart (Neon Pink)
ax2.plot(year_df['release_year'], year_df['duration_min'], color='#FF007F', marker='s', linewidth=2.2, markersize=5, linestyle='--')
ax2.fill_between(year_df['release_year'], year_df['duration_min'], color='#FF007F', alpha=0.25)
ax2.set_title('Average Song Duration Over Time (Minutes)', fontsize=10, fontweight='bold', color='white', pad=8)
ax2.set_xlabel('Release Year', fontsize=9, fontweight='bold', color='white')
ax2.set_ylabel('Duration (Mins)', fontsize=8.5, color='#FF007F', fontweight='bold')
ax2.grid(True, linestyle=':', alpha=0.25, color='gray')
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)

plt.tight_layout()

# Save image
plt.savefig('vibrant_year_trends.png', dpi=300, facecolor='#121212')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid", palette="pastel")

# Load your clean dataset
df = pd.read_csv('spotify_cleaned_sample.csv')
plt.figure(figsize=(10, 5))
genre_means = df.groupby('playlist_genre')[['danceability', 'energy', 'valence', 'acousticness']].mean().reset_index()
df_melted = pd.melt(genre_means, id_vars=['playlist_genre'], var_name='Audio Feature', value_name='Average Score')

sns.barplot(data=df_melted, x='playlist_genre', y='Average Score', hue='Audio Feature', palette='mako')
plt.title('What Makes Each Genre Unique? (Audio Traits Comparison)', fontsize=12, fontweight='bold', pad=12)
plt.xlabel('Music Genre', fontsize=10)
plt.ylabel('Average Score (0.0 to 1.0)', fontsize=10)
plt.legend(title='Feature', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.savefig('chart_genre_traits.png', dpi=300)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load data
df = pd.read_excel('spotify_cleaned_sample.xlsx')

# 2. Get Top 10 Artists by Total Duration
top_artist_mins = df.groupby('track_artist')['duration_min'].sum().nlargest(10).reset_index()

# 3. Plot using Seaborn (clean white background, no grid)
plt.figure(figsize=(6, 3), dpi=300)
sns.set_style("white")

ax = sns.barplot(
    data=top_artist_mins,
    y='track_artist',
    x='duration_min',
    palette='viridis'
)

plt.title('Top Artists by Total Playtime (Minutes)', fontsize=10, fontweight='bold', pad=10)
plt.xlabel('Total Duration (Minutes)', fontsize=8, fontweight='bold')
plt.ylabel('Artist Name', fontsize=8, fontweight='bold')
plt.tick_params(axis='both', labelsize=8)
sns.despine()

# 4. Add numbers inside the bars
for p in ax.patches:
    width = p.get_width()
    ax.annotate(f'{width:.1f}m', 
                (width / 2, p.get_y() + p.get_height() / 2.),
                ha='center', va='center', fontsize=8, color='white', weight='bold')

plt.tight_layout()
plt.savefig('top_artists_playtime_sns.png', dpi=300)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load dataset
df = pd.read_excel('spotify_cleaned_sample.xlsx')

# 2. Setup 1x2 Figure Layout
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5), dpi=300)
sns.set_style("white")

# Panel 1: Mood vs Energy (Valence vs Energy)
sns.scatterplot(
    data=df,
    x='valence',
    y='energy',
    hue='playlist_genre',
    size='track_popularity',
    palette='viridis',
    sizes=(30, 180),
    alpha=0.85,
    ax=ax1
)
ax1.set_title('1. Mood vs. Energy Quadrants', fontsize=10.5, fontweight='bold', pad=10)
ax1.set_xlabel('Valence / Mood Score (0.0 to 1.0)', fontsize=8.5, fontweight='bold')
ax1.set_ylabel('Energy Level (0.0 to 1.0)', fontsize=8.5, fontweight='bold')
ax1.tick_params(axis='both', labelsize=8)

# Panel 1 Reference lines
ax1.axvline(x=0.5, color='gray', linestyle='--', alpha=0.4)
ax1.axhline(y=0.5, color='gray', linestyle='--', alpha=0.4)

ax1.text(0.08, 0.93, '[Intense / Dark]', fontsize=7.5, fontweight='bold', color='#B22222')
ax1.text(0.68, 0.93, '[Party / Upbeat]', fontsize=7.5, fontweight='bold', color='#008000')
ax1.text(0.08, 0.12, '[Sad / Acoustic]', fontsize=7.5, fontweight='bold', color='#4682B4')
ax1.text(0.68, 0.12, '[Chill / Happy]', fontsize=7.5, fontweight='bold', color='#CC8800')
ax1.get_legend().remove()  # Remove duplicate legend
sns.despine(ax=ax1)

# Panel 2: Production Matrix (Tempo vs Loudness)
scatter = sns.scatterplot(
    data=df,
    x='tempo',
    y='loudness',
    hue='playlist_genre',
    size='track_popularity',
    palette='flare',
    sizes=(30, 200),
    alpha=0.85
)

plt.title('Audio Production Matrix: Tempo vs. Loudness Profile', fontsize=10.5, fontweight='bold', pad=10)
plt.xlabel('Tempo (BPM)', fontsize=8.5, fontweight='bold')
plt.ylabel('Loudness (dB)', fontsize=8.5, fontweight='bold')
plt.tick_params(axis='both', labelsize=8)

# 3. Dynamic Midpoints based on YOUR exact dataset medians
med_tempo = df['tempo'].median()
med_loudness = df['loudness'].median()

plt.axvline(x=med_tempo, color='gray', linestyle='--', alpha=0.5)
plt.axhline(y=med_loudness, color='gray', linestyle='--', alpha=0.5)

# 4. Zoom Y-axis so dots fill the full graph canvas
plt.ylim(df['loudness'].min() - 0.8, df['loudness'].max() + 1.2)

# 5. Position text relative to actual min/max
plt.text(df['tempo'].min() + 2, df['loudness'].max() - 0.8, '[Mid Tempo / High Loudness]', fontsize=7.5, fontweight='bold', color='#B22222')
plt.text(med_tempo + 3, df['loudness'].max() - 0.8, '[Fast Tempo / Peak Loudness]', fontsize=7.5, fontweight='bold', color='#008000')
plt.text(df['tempo'].min() + 2, df['loudness'].min() + 0.3, '[Moderate Loudness]', fontsize=7.5, fontweight='bold', color='#4682B4')
plt.text(med_tempo + 3, df['loudness'].min() + 0.3, '[Fast / Dynamic Range]', fontsize=7.5, fontweight='bold', color='#CC8800')

plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', fontsize=7.5, title='Genre & Popularity', title_fontsize=8)
sns.despine()

plt.tight_layout()
plt.savefig('actual_clustered_loudness.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Read dataset
df = pd.read_csv('spotify_cleaned_sample.csv')

# 2. Get the lowest 10 performing tracks
lowest_10 = df.sort_values(by='track_popularity').head(10).copy()

# 3. Clean up and shorten track names
def clean_title(name):
    name = name.split(' - ')[0]  # Remove version info
    if len(name) > 15:
        return name[:12] + '...'
    return name

lowest_10['short_name'] = lowest_10['track_name'].apply(clean_title)

# 4. Plot line graph
plt.figure(figsize=(10, 4.5))
plt.plot(lowest_10['short_name'], lowest_10['danceability'], marker='o', label='Danceability', color='blue')
plt.plot(lowest_10['short_name'], lowest_10['energy'], marker='s', label='Energy', color='orange')

# 5. Format labels
plt.title('Danceability and Energy of Lowest 10 Tracks')
plt.xlabel('Track Name')
plt.ylabel('Score (0 to 1)')
plt.xticks(rotation=30, ha='right')  # Rotate for clean alignment
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.savefig('Lowest performers', dpi=300)
plt.show()
```
